## 1. Why strings matter in cyber

In cybersecurity, a lot of data comes as **text**:
- logs
- URLs
- IP addresses
- HTTP requests
- usernames
- alerts

That means string operations are a core skill.

Common operations:
- `split()` → break text into parts
- `lower()` / `upper()` → normalize text
- `startswith()` / `endswith()` → check patterns
- `in` → search inside text
- `replace()` → modify text


````
# Example
log_line = "192.168.1.10 - GET /login HTTP/1.1"

print(log_line.split())
print("login" in log_line)
print(log_line.startswith("192"))
print(log_line.replace("GET", "POST"))
````

## 2. Extracting parts from log lines

A typical log line can contain:
- source IP
- HTTP method
- endpoint
- protocol

Example:

`192.168.1.10 - GET /login HTTP/1.1`

After splitting:
- index `0` → IP
- index `2` → method
- index `3` → endpoint
- index `4` → protocol


## Exercise 2
Using the line below, extract and print:
- IP address
- method
- endpoint

```python
line = "172.16.0.7 - GET /dashboard HTTP/1.1"
```


````
# Solution 2
line = "172.16.0.7 - GET /dashboard HTTP/1.1"
parts = line.split()

ip = parts[0]
method = parts[2]
endpoint = parts[3]

print("IP:", ip)
print("Method:", method)
print("Endpoint:", endpoint)
````

## 4. Filtering suspicious activity

A simple way to detect suspicious behavior is to look for:
- `/admin`
- `FAILED_LOGIN`
- unusual HTTP methods
- repeated requests from the same IP

Filtering means keeping only the lines that match a condition.


## Exercise 4
From the list below, print only the lines that contain `/admin`.


````
# Solution 4
logs = [
    "192.168.1.10 - GET /login HTTP/1.1",
    "10.0.0.8 - POST /admin HTTP/1.1",
    "172.16.0.3 - GET /home HTTP/1.1",
    "192.168.1.50 - GET /admin HTTP/1.1"
]

for log in logs:
    if "/admin" in log:
        print(log)
````

## 5. List comprehensions for filtering

A list comprehension is a compact way to build a new list.

Example:
```python
admin_logs = [log for log in logs if "/admin" in log]
```

This is very useful in log analysis.


## Exercise 5
Use a list comprehension to create a new list called `failed_logs` that contains only lines with `FAILED_LOGIN`.


````
# Solution 5
logs = [
    "192.168.1.10 FAILED_LOGIN",
    "192.168.1.11 SUCCESS_LOGIN",
    "10.0.0.8 FAILED_LOGIN",
    "172.16.0.3 SUCCESS_LOGIN"
]

failed_logs = [log for log in logs if "FAILED_LOGIN" in log]
print(failed_logs)
````

## 6. Counting suspicious events

Counting is one of the most common analysis tasks.

We may want to count:
- how many failed logins happened
- how many `/admin` requests exist
- how many times an IP appears


## Exercise 6
Count how many lines contain `FAILED_LOGIN`.


````
# Solution 6
logs = [
    "192.168.1.10 FAILED_LOGIN",
    "192.168.1.11 SUCCESS_LOGIN",
    "10.0.0.8 FAILED_LOGIN",
    "172.16.0.3 SUCCESS_LOGIN",
    "10.0.0.8 FAILED_LOGIN"
]

count = 0
for log in logs:
    if "FAILED_LOGIN" in log:
        count += 1

print("Failed login count:", count)
````

## 7. Dictionaries for counting per IP

A dictionary is great when we want to count occurrences by key.

Example:
- key → IP address
- value → number of failed attempts


## Exercise 7
Count the number of failed logins **per IP**.

Expected idea:
- extract the IP
- if the line contains `FAILED_LOGIN`, update the dictionary


````
# Solution 7
logs = [
    "192.168.1.10 FAILED_LOGIN",
    "192.168.1.10 FAILED_LOGIN",
    "10.0.0.8 FAILED_LOGIN",
    "172.16.0.3 SUCCESS_LOGIN",
    "10.0.0.8 FAILED_LOGIN",
    "10.0.0.8 FAILED_LOGIN"
]

failed_per_ip = {}

for log in logs:
    if "FAILED_LOGIN" in log:
        ip = log.split()[0]
        if ip in failed_per_ip:
            failed_per_ip[ip] += 1
        else:
            failed_per_ip[ip] = 1

print(failed_per_ip)
````

## 8. Detecting brute-force behavior

A very simple brute-force indicator:
- the same IP has many failed logins

This is not a full detection system, but it is a good beginner exercise.


## Exercise 8
Using the dictionary logic, print only the IPs that have **3 or more** failed logins.


````
# Solution 8
logs = [
    "192.168.1.10 FAILED_LOGIN",
    "192.168.1.10 FAILED_LOGIN",
    "192.168.1.10 FAILED_LOGIN",
    "10.0.0.8 FAILED_LOGIN",
    "10.0.0.8 FAILED_LOGIN",
    "172.16.0.3 SUCCESS_LOGIN"
]

failed_per_ip = {}

for log in logs:
    if "FAILED_LOGIN" in log:
        ip = log.split()[0]
        failed_per_ip[ip] = failed_per_ip.get(ip, 0) + 1

for ip, count in failed_per_ip.items():
    if count >= 3:
        print(ip, "is suspicious with", count, "failed logins")
````

## 11. Parsing structured HTTP logs

Now let’s work with HTTP-style lines such as:

`192.168.1.5 - POST /login HTTP/1.1`

We can convert each line into a dictionary:
```python
{
    "ip": "...",
    "method": "...",
    "endpoint": "..."
}
```

This is much easier to analyze later.


## Exercise 11
Convert each HTTP log line into a dictionary with keys:
- `ip`
- `method`
- `endpoint`

Store the dictionaries in a list called `parsed_logs`.


````
# Solution 11
logs = [
    "192.168.1.5 - POST /login HTTP/1.1",
    "10.0.0.8 - GET /home HTTP/1.1",
    "172.16.0.4 - GET /admin HTTP/1.1"
]

parsed_logs = []

for log in logs:
    parts = log.split()
    entry = {
        "ip": parts[0],
        "method": parts[2],
        "endpoint": parts[3]
    }
    parsed_logs.append(entry)

print(parsed_logs)
````

## 12. Counting requests to sensitive endpoints

Sensitive endpoints may include:
- `/admin`
- `/login`
- `/config`
- `/backup`

We often want to know how many times they were accessed.


## Exercise 12
Using `parsed_logs`, count how many requests were made to `/admin`.


````
# Solution 12
logs = [
    "192.168.1.5 - POST /login HTTP/1.1",
    "10.0.0.8 - GET /admin HTTP/1.1",
    "172.16.0.4 - GET /admin HTTP/1.1",
    "10.0.0.9 - GET /home HTTP/1.1"
]

parsed_logs = []

for log in logs:
    parts = log.split()
    parsed_logs.append({
        "ip": parts[0],
        "method": parts[2],
        "endpoint": parts[3]
    })

admin_count = 0
for entry in parsed_logs:
    if entry["endpoint"] == "/admin":
        admin_count += 1

print("Requests to /admin:", admin_count)
````

## 13. Finding unique IP addresses

Another common task:
- find all unique IPs
- count how many unique clients connected

A `set` is useful for this.


## Exercise 13
Find the unique IP addresses from the HTTP logs.


````
# Solution 13
logs = [
    "192.168.1.5 - POST /login HTTP/1.1",
    "10.0.0.8 - GET /admin HTTP/1.1",
    "172.16.0.4 - GET /admin HTTP/1.1",
    "10.0.0.8 - GET /home HTTP/1.1"
]

unique_ips = set()

for log in logs:
    ip = log.split()[0]
    unique_ips.add(ip)

print(unique_ips)
print("Unique IP count:", len(unique_ips))
````

## 14. Mixed log analysis mini-project

Now we combine:
- strings
- loops
- conditions
- dictionaries
- files

### Goal
Read a file and produce:
1. total failed logins  
2. failed logins per IP  
3. suspicious IPs with 3 or more failed logins


## Exercise 14
Create a file called `auth_logs.txt` with these lines:

```text
192.168.1.10 FAILED_LOGIN
192.168.1.10 FAILED_LOGIN
192.168.1.10 FAILED_LOGIN
10.0.0.8 FAILED_LOGIN
10.0.0.8 SUCCESS_LOGIN
172.16.0.3 FAILED_LOGIN
```

Then analyze the file and print:
- total failed logins
- dictionary of failed logins per IP
- suspicious IPs (3 or more failures)


````
# Solution 14
auth_logs = [
    "192.168.1.10 FAILED_LOGIN\n",
    "192.168.1.10 FAILED_LOGIN\n",
    "192.168.1.10 FAILED_LOGIN\n",
    "10.0.0.8 FAILED_LOGIN\n",
    "10.0.0.8 SUCCESS_LOGIN\n",
    "172.16.0.3 FAILED_LOGIN\n"
]

with open("auth_logs.txt", "w") as f:
    f.writelines(auth_logs)

with open("auth_logs.txt", "r") as f:
    lines = f.readlines()

total_failed = 0
failed_per_ip = {}

for line in lines:
    line = line.strip()
    if "FAILED_LOGIN" in line:
        total_failed += 1
        ip = line.split()[0]
        failed_per_ip[ip] = failed_per_ip.get(ip, 0) + 1

print("Total failed logins:", total_failed)
print("Failed per IP:", failed_per_ip)

print("Suspicious IPs:")
for ip, count in failed_per_ip.items():
    if count >= 3:
        print("-", ip, "with", count, "failed attempts")
````

## 15. Optional challenge

Try solving this on your own:

### Challenge
Given HTTP logs, find:
- total number of `GET` requests
- total number of `POST` requests
- all requests to `/admin`
- IPs that visited `/login`

This is a good checkpoint before moving to:
- regular expressions
- JSON
- CSV
- packet/log parsing


````
# Possible challenge solution
logs = [
    "192.168.1.5 - POST /login HTTP/1.1",
    "10.0.0.8 - GET /admin HTTP/1.1",
    "172.16.0.4 - GET /admin HTTP/1.1",
    "10.0.0.8 - GET /home HTTP/1.1",
    "192.168.1.7 - POST /login HTTP/1.1"
]

get_count = 0
post_count = 0
admin_requests = []
login_ips = set()

for log in logs:
    parts = log.split()
    ip = parts[0]
    method = parts[2]
    endpoint = parts[3]

    if method == "GET":
        get_count += 1
    elif method == "POST":
        post_count += 1

    if endpoint == "/admin":
        admin_requests.append(log)

    if endpoint == "/login":
        login_ips.add(ip)

print("GET requests:", get_count)
print("POST requests:", post_count)
print("Admin requests:", admin_requests)
print("IPs visiting /login:", login_ips)
````
