## Exercise 5

**Request:**

Given a dictionary of failed login attempts per user, find which user has the highest number of failed attempts.

```python
failed_logins = {
    "admin": 5,
    "student1": 2,
    "guest": 7,
    "teacher": 3
}

max_fails = 0
failed_user= ""

for role, count in failed_logins.items():
    if count > max_fails:
        max_fails = count
        failed_user = role

print("Most failed role login: " + failed_user
      + " with number of fails: " + str(max_fails))


```

## Exercise 10

**Request:**

Create a function `count_failed_attempts(attempts)` that takes a list like `[True, False, True]` where `True` means failed login, and returns how many failed attempts there are.

```python
def count_failed_attempts(attempts):
    count = 0
    for attempt in attempts:
        if attempt is True:
            count+=1
    return count
```

```python
attempts = [True, True, False]
count_failed_attempts(attempts)
```

```python
def count_failed_attempts_short_way(attempts):
    return sum(attempts)
```

## Exercise 12 — Calculate Network Packet Size

Create a function `packet_size(data_size, header_size, encryption_overhead)`.

The function should return the **total packet size**.

```python
def packet_size(data_size, header_size, encryption_overhead):
    return data_size+header_size+encryption_overhead
```

## Example — Functions with Arbitrary Parameters (*args)

Sometimes we do not know how many parameters will be passed to a function.

Using `*args` allows the function to accept multiple values.

```python
# sum of numbers
def sum_numbers(*numbers):
    sum = 0
    for num in numbers:
        sum += num
    return sum

# short way
# def sum_numbers(*numbers):
#     return sum(numbers)
```

```python
sum_numbers(1,2,3,4)
```

## Example — Keyword Arguments (**kwargs)

`**kwargs` allows a function to receive named parameters.

```python
# show user info username, role etc

def show_user_info(**infos):
    for info in infos.items():
        print(f"{info[0]} : {info[1]}")
        # print(info)
```

## Cybersecurity Example — Log Event (**kwargs)

Create a function that prints security log information such as IP address, action, and status.

```python
def print_security_log(**logs):
    for key, value in logs.items():
        print(key,value)
```

```python
print_security_log(ip_address = '192.168.1.0', action ='read', status ='active', failed = True)
```

```python
print_security_log(ip_address = '192.168.1.0', action ='read')
```
