8. Given a list of words, count how many words have more than 5 characters.

```python
words = ['python', 'Math', 'Programming', 'Cyber Security']

counter = 0

for word in words:
    if len(word) > 5:
        counter+=1

print(counter)
```

```python
# print words with more than 7 chars and start p
words = ['python', 'Math', 'Programming', 'Cyber Security']

new_words = []

for word in words:
    if len(word)>7 and word[0].lower() =='p':
        new_words.append(word)

print(new_words)
```

10. Given a list of numbers, count how many numbers are positive, negative, and zero.

```python
lista_2 = [10, -10, 2, 0, 5, -5, 0]

counter_pos= 0
counter_neg = 0
counter_zero = 0

for element in lista_2:
    if element > 0:
        counter_pos += 1
    elif element < 0:
        counter_neg += 1
    else:
        counter_zero +=1

print("Positive numbers count: ", counter_pos)
print("Negative numbers count: ", counter_neg)
print("Zero numbers count: ", counter_zero)

```

11. Given a list of numbers, create a new list that removes duplicate values.

```python
lista_3 = [2, 2 , 3 , 3 , 4 , 4 ,5]
new_list = []

for element in lista_3:
    if element not in new_list:
        new_list.append(element)

print(new_list)
```

12. Given a list of numbers, reverse the list using a loop (do not use reverse() or slicing).

```python
lista_4 = [10,20,30,40,50,60,70]
last_index = len(lista_4) - 1  #6
reversed_list = []

for index in range(last_index, -1, -1):
    reversed_list.append(lista_4[index])

print(reversed_list)

```

```python
# 12 b)
reversed_list = []
last_index = len(lista_4) - 1

while last_index >=0:
    reversed_list.append(lista_4[last_index])
    last_index -= 1
```
