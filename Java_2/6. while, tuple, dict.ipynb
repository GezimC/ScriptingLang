```python
# 16. Create a matrix (3,3), find max value and its position

matrix = [
    [2,5,10],
    [7,1,8],
    [12, 1, 5]
]
```

```python
matrix
```

```python
for element in matrix:
    print(element)
```

```python
temp = matrix[0][0]  # max = 2
index_of_max = [0,0]
for i in range(0,3):
    for j in range(0,3):
        if temp < matrix[i][j]:
            temp =  matrix[i][j]
            index_of_max[0] = i
            index_of_max[1] = j

print(temp, index_of_max)
```

```python
# print random number until 3 appears

import random as r
r.seed(101)
```
