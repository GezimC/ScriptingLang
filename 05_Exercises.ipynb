## Basic Exercise 1 — map()

numbers = [1, 2, 3, 4, 5]

Use `map()` to create a new list where each number is multiplied by 3.


```
numbers = [1,2,3,4,5]
result = []
for num in numbers:
    result.append(num*3)

print(result)
```

```
def multply_by_three(number):
    return number*3
```

```
list(map(multply_by_three, numbers))
```

```
# lambda function - funksion anonim

list(map(lambda num: num*3  , numbers))
```

```
arr_1 = [1,2,3,4,5]
arr_2 = [5,4,3,2,1]

sum = []

for index in range(0, len(arr_1)):
    sum.append(arr_1[index] + arr_2[index])

print(sum)
```

```
list(map(lambda num1,num2: num1+num2  , arr_1, arr_2  ))
```

```
# list comprehension

numbers

result_3 = [num*3 for num in numbers]
```

```
print(numbers)
print(result_3)
```

## Basic Exercise 2 — filter()

numbers = [10, 15, 20, 25, 30]

Use `filter()` to keep only numbers greater than 18.


```
def filter_larger_18(num):
    if num > 18:
        return True
    else:
        return False
```

```
numbers_1 = [10,15,20,25,30]

list(filter( filter_larger_18 , numbers_1))
```

```
[num for num in numbers_1 if num>18]
```

```
[num>18 for num in numbers_1]
```

## Basic Exercise 5 — list comprehension with if-else

numbers = [1, 2, 3, 4, 5]

Create a new list where:
- "even" is stored if the number is even
- "odd" is stored if the number is odd


```
number = 10

if number % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```
print('Even')  if number%2 ==0  else print('Odd')
```

```
numbers_3 = [1,2,3,4,5,6,8]

['Even' if num % 2 == 0 else 'Odd' for num in numbers_3]
```

```
[pow(num,2)  if num%2==0 else pow(num,3) for num in numbers_3]
```

- "secure" → HTTPS, SSH
