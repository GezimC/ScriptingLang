## What is JSON?

JSON stands for **JavaScript Object Notation**.

In Python:
- a JSON **object** becomes a `dict`
- a JSON **array** becomes a `list`
- JSON strings, numbers, booleans, and null map to Python types

You will often see JSON in cyber security when working with:
- API responses
- SIEM exports
- authentication events
- threat intelligence feeds
- cloud logs

```python

import json
```

## 1. JSON from a string
A small login event in JSON format:

```python

login_event_json = '''
{
    "timestamp": "2026-04-08T08:30:00",
    "username": "alice",
    "ip_address": "185.23.44.10",
    "country": "Germany",
    "status": "FAILED",
    "device": "Windows 11"
}
'''

login_event = json.loads(login_event_json)

print(login_event)
print(type(login_event))
```

### Accessing values inside JSON

```python

print("Username:", login_event["username"])
print("IP address:", login_event["ip_address"])
print("Status:", login_event["status"])
```

## 2. Nested JSON example
Real logs are often more complex and nested.

```python

alert_json = '''
{
    "alert_id": 501,
    "severity": "high",
    "source": {
        "ip": "45.77.19.2",
        "country": "Unknown"
    },
    "event": {
        "type": "multiple_failed_logins",
        "count": 12
    },
    "tags": ["authentication", "bruteforce", "priority"]
}
'''

alert = json.loads(alert_json)

print("Severity:", alert["severity"])
print("Source IP:", alert["source"]["ip"])
print("Event type:", alert["event"]["type"])
print("First tag:", alert["tags"][0])
```

## 3. JSON array of events
A JSON array becomes a Python list.

```python

events_json = '''
[
    {"username": "alice", "status": "FAILED", "ip": "10.0.0.5"},
    {"username": "bob", "status": "SUCCESS", "ip": "10.0.0.6"},
    {"username": "alice", "status": "FAILED", "ip": "10.0.0.5"}
]
'''

events = json.loads(events_json)

print(type(events))
print("Number of events:", len(events))
print("First event:", events[0])
```

### Loop through events

```python

for event in events:
    print(event["username"], "-", event["status"], "-", event["ip"])
```

## 4. Filtering suspicious JSON events
We will keep only failed login attempts.

```python

failed_events = []

for event in events:
    if event["status"] == "FAILED":
        failed_events.append(event)

print("Failed events:")
for event in failed_events:
    print(event)
```

## 5. Counting failed logins per user from JSON

```python

failed_per_user = {}

for event in events:
    username = event["username"]
    if event["status"] == "FAILED":
        if username not in failed_per_user:
            failed_per_user[username] = 0
        failed_per_user[username] += 1

print(failed_per_user)
```

## 6. Save JSON to a file, then read it back
This is useful because in real work you usually get files, not just strings.

```python

sample_json_events = [
    {"timestamp": "2026-04-08T08:10:00", "username": "alice", "status": "FAILED", "ip": "10.0.0.5"},
    {"timestamp": "2026-04-08T08:11:00", "username": "alice", "status": "FAILED", "ip": "10.0.0.5"},
    {"timestamp": "2026-04-08T08:12:00", "username": "bob", "status": "SUCCESS", "ip": "10.0.0.6"},
    {"timestamp": "2026-04-08T08:13:00", "username": "carol", "status": "FAILED", "ip": "10.0.0.7"}
]

with open("login_events.json", "w", encoding="utf-8") as f:
    json.dump(sample_json_events, f, indent=4)

print("File login_events.json created.")
```

```python

with open("login_events.json", "r", encoding="utf-8") as f:
    events_from_file = json.load(f)

print("Events loaded from file:")
for event in events_from_file:
    print(event)
```

# CSV Section

CSV stands for **Comma-Separated Values**.

It is very common in cyber security because many tools export data in CSV format:
- login reports
- vulnerability scan results
- email reports
- network connection summaries
- ticket exports

```python

import csv
```

## 7. Create a small CSV file

```python

csv_rows = [
    ["timestamp", "username", "ip_address", "status", "country"],
    ["2026-04-08 08:00:00", "alice", "185.23.44.10", "FAILED", "Germany"],
    ["2026-04-08 08:01:00", "bob", "91.198.10.4", "SUCCESS", "Kosovo"],
    ["2026-04-08 08:02:00", "alice", "185.23.44.10", "FAILED", "Germany"],
    ["2026-04-08 08:03:00", "carol", "77.45.21.9", "FAILED", "France"]
]

with open("login_report.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(csv_rows)

print("File login_report.csv created.")
```

## 8. Read CSV row by row

```python

with open("login_report.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

## 9. Read CSV using `DictReader`
`DictReader` is easier because each row becomes a dictionary.

```python

with open("login_report.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row)
```

## 10. Filter failed logins from CSV

```python

failed_rows = []

with open("login_report.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        if row["status"] == "FAILED":
            failed_rows.append(row)

print("Failed rows:")
for row in failed_rows:
    print(row)
```

## 11. Count failed attempts per IP from CSV

```python
import csv
failed_per_ip = {}

with open("login_report.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        ip = row["ip_address"]
        if row["status"] == "FAILED":
            if ip not in failed_per_ip:
                failed_per_ip[ip] = 0
            failed_per_ip[ip] += 1

print(failed_per_ip)
```

## 12. Export suspicious rows to a new CSV

```python

with open("failed_logins_only.csv", "w", newline="", encoding="utf-8") as f:
    fieldnames = ["timestamp", "username", "ip_address", "status", "country"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(failed_rows)

print("File failed_logins_only.csv created.")
```

## 14. Load the mini project data

```python

with open("suspicious_logins.json", "r", encoding="utf-8") as f:
    project_events = json.load(f)

print("Total loaded events:", len(project_events))
print("First event:", project_events[0])
```

## 15. Total logins and failed logins

```python

total_logins = len(project_events)

failed_logins = 0
for event in project_events:
    if event["status"] == "FAILED":
        failed_logins += 1

print("Total logins:", total_logins)
print("Failed logins:", failed_logins)
```

## 16. Failed attempts per user

```python

failed_per_user = {}

for event in project_events:
    if event["status"] == "FAILED":
        username = event["username"]
        if username not in failed_per_user:
            failed_per_user[username] = 0
        failed_per_user[username] += 1

print("Failed attempts per user:")
for user, count in failed_per_user.items():
    print(user, "->", count)
```

## 17. Failed attempts per IP

```python

failed_per_ip = {}

for event in project_events:
    if event["status"] == "FAILED":
        ip = event["ip_address"]
        if ip not in failed_per_ip:
            failed_per_ip[ip] = 0
        failed_per_ip[ip] += 1

print("Failed attempts per IP:")
for ip, count in failed_per_ip.items():
    print(ip, "->", count)
```

## 18. Detect logins outside working hours

For this notebook, assume working hours are **08:00 to 18:00**.
Any event before 08:00 or after 18:00 will be flagged.

```python

outside_working_hours = []

for event in project_events:
    hour = int(event["timestamp"][11:13])
    if hour < 8 or hour >= 18:
        outside_working_hours.append(event)

print("Events outside working hours:", len(outside_working_hours))
for event in outside_working_hours:
    print(event["timestamp"], "-", event["username"], "-", event["country"], "-", event["status"])
```

## 19. Detect users seen from multiple countries
A normal user is usually expected from one country or a small known set.
Here we simply detect more than one country in the data.

```python

countries_per_user = {}

for event in project_events:
    username = event["username"]
    country = event["country"]

    if username not in countries_per_user:
        countries_per_user[username] = set()

    countries_per_user[username].add(country)

print("Countries per user:")
for user, countries in countries_per_user.items():
    print(user, "->", countries)
```

```python

multi_country_users = []

for user, countries in countries_per_user.items():
    if len(countries) > 1:
        multi_country_users.append(user)

print("Users seen from multiple countries:", multi_country_users)
```

## 20. Create a suspicious events list
We will flag an event as suspicious if **at least one** of these is true:
- login failed
- outside working hours
- user seen from multiple countries

```python

suspicious_events = []

for event in project_events:
    username = event["username"]
    hour = int(event["timestamp"][11:13])

    is_failed = event["status"] == "FAILED"
    is_outside_hours = hour < 8 or hour >= 18
    is_multi_country_user = username in multi_country_users

    if is_failed or is_outside_hours or is_multi_country_user:
        suspicious_events.append(event)

print("Suspicious events found:", len(suspicious_events))
for event in suspicious_events[:10]:
    print(event)
```

## 21. Export suspicious events to CSV

```python

with open("suspicious_events_output.csv", "w", newline="", encoding="utf-8") as f:
    fieldnames = ["timestamp", "username", "ip_address", "country", "status", "device"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(suspicious_events)

print("File suspicious_events_output.csv created.")
```

## 22. Simple incident summary
A script should not only process data, but also produce a short summary for an analyst.

```python

top_user = None
top_user_count = 0

for user, count in failed_per_user.items():
    if count > top_user_count:
        top_user = user
        top_user_count = count

top_ip = None
top_ip_count = 0

for ip, count in failed_per_ip.items():
    if count > top_ip_count:
        top_ip = ip
        top_ip_count = count

print("INCIDENT SUMMARY")
print("-" * 50)
print("Total logins:", total_logins)
print("Total failed logins:", failed_logins)
print("Most failed user:", top_user, "with", top_user_count, "failed attempts")
print("Most suspicious IP by failures:", top_ip, "with", top_ip_count, "failed attempts")
print("Outside working hours:", len(outside_working_hours), "events")
print("Users seen from multiple countries:", ", ".join(multi_country_users) if multi_country_users else "None")
print("Suspicious events exported to: suspicious_events_output.csv")
```
