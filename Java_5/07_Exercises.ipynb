## What is JSON?

JSON stands for **JavaScript Object Notation**.

In Python:
- a JSON **object** becomes a `dict`
- a JSON **array** becomes a `list`
- JSON strings, numbers, booleans, and null map to Python types

You will often see JSON in cyber security when working with:
- API responses
- SIEM exports
- authentication events
- threat intelligence feeds
- cloud logs

```python
import json
from pickle import DICT
```

## 3. JSON array of events

### Loop through events

```python
# print failed events
for event in events_dict:
    if event['status'] == 'FAILED':
        print(f'Username: {event['username']} and ip is: {event['ip']} ')
```

# CSV Section

CSV stands for **Comma-Separated Values**.

It is very common in cyber security because many tools export data in CSV format:
- login reports
- vulnerability scan results
- email reports
- network connection summaries
- ticket exports

```python
import csv
```

## 7. Create a small CSV file

```python
import csv
csv_rows = [
    ["timestamp", "username", "ip_address", "status", "country"],
    ["2026-04-08 08:00:00", "alice", "185.23.44.10", "FAILED", "Germany"],
    ["2026-04-08 08:01:00", "bob", "91.198.10.4", "SUCCESS", "Kosovo"],
    ["2026-04-08 08:02:00", "alice", "185.23.44.10", "FAILED", "Germany"],
    ["2026-04-08 08:03:00", "carol", "77.45.21.9", "FAILED", "France"]
]

with open("csv_rows.csv", "w", newline="") as file:
     writer = csv.writer(file)
     writer.writerows(csv_rows)


#csv.writerfile()
#write.writerows(csv_rows)
```

## 8. Read CSV row by row

```python
import csv

with open("csv_rows.csv", 'r', newline="") as file:
    reader = csv.reader(file)
    new_list = list(reader)


print(new_list)
```

## 9. Read CSV using `DictReader`
`DictReader` is easier because each row becomes a dictionary.

```python
all_events = []
with open("csv_rows.csv", 'r', newline="" ) as file:
    reader = csv.DictReader(file)
    for row in reader:
        all_events.append(row)

```

```python
all_events
```

## 10. Filter failed logins from CSV

```python
failed_rows = []

for event in all_events:
    if event['status'] == 'FAILED':
        failed_rows.append(event)


print(failed_rows)
```

## 11. Count failed attempts per IP from CSV

```python

failed_per_ip = {}

for event in failed_rows:
    ip = event['ip_address']
    if ip in failed_per_ip:
        failed_per_ip[ip] += 1
    else:
        failed_per_ip[ip] = 1


print(failed_per_ip)
```

## 12. Export suspicious rows to a new CSV

```python
failed_rows[0].keys()
```

```python

with open("failed_logins_only.csv", "w", newline="") as f:
    fieldnames = ["timestamp", "username", "ip_address", "status", "country"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(failed_rows)

print("File failed_logins_only.csv created.")
```

```python
failed_per_user = {}

for login in suspicious_logins:
    user = login['username']
    if user in failed_per_user:
        failed_per_user[user] += 1
    else:
        failed_per_user[user] = 1

print(failed_per_user)
```

## 17. Failed attempts per IP

```python
failed_per_ipadd = {}

for login in suspicious_logins:
    ip_add = login['ip_address']
    if ip_add in failed_per_ipadd:
        failed_per_ipadd[ip_add] += 1
    else:
        failed_per_ipadd[ip_add] = 1

print(failed_per_ipadd)
```

## 18. Detect logins outside working hours

For this notebook, assume working hours are **08:00 to 18:00**.
Any event before 08:00 or after 18:00 will be flagged.

```python
suspicious_logins
```

```python
event_5 = suspicious_logins[5]

time_1 = event_5['timestamp'].split('T')[1]

print(time_1)
```

```python
int(time_1[0:2]) < 8 or int(time_1[0:2]) > 18
```

```python
outside_working_hours = []

for login in suspicious_logins:
    timestamp = login['timestamp'].split('T')[1]
    hour = int(timestamp[0:2])

    if hour < 8 or hour > 18:
        outside_working_hours.append(login)

for event in outside_working_hours:
    print(f"Username: {event['username']}, timestamp: {event['timestamp']}")


```

## 19. Detect users seen from multiple countries
A normal user is usually expected from one country or a small known set.
Here we simply detect more than one country in the data.

```python
suspicious_logins[0]
```

```python
countries_per_user = {}

for event in suspicious_logins:
    user = event['username']
    country = event['country']

    if user not in countries_per_user.keys():
        countries_per_user[user] = set()

    countries_per_user[user].add(country)


print(countries_per_user)
```

```python

multi_country_users = []

for user in countries_per_user:
    if len(countries_per_user[user]) > 1:
        multi_country_users.append(user)


print(multi_country_users)
```

## 20. Create a suspicious events list
We will flag an event as suspicious if **at least one** of these is true:
- login failed
- outside working hours
- user seen from multiple countries

```python
suspicious_logins[0]
```

```python
print(multi_country_users)
```

```python

suspicious_events = []

for event in suspicious_logins:

    isFailed = event['status'] == 'FAILED'  #true

    hour = int(event['timestamp'].split('T')[1][0:2])

    isOutsideHours = hour < 8 or hour > 18  #false

    isMultipleCountries = event['username'] in multi_country_users #false

    if isFailed or isOutsideHours or isMultipleCountries:
        suspicious_events.append(event)


print(suspicious_events)

```

## 21. Export suspicious events to CSV

```python
suspicious_events
```

```python
import csv
with open("exported_results.csv", "w", newline="") as f:
    fieldnames = ["timestamp", "username", "ip_address", "status", "country", "device"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(suspicious_events)

print("File exported_results.csv created.")
```
