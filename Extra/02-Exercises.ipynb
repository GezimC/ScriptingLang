# Python List and Loop Exercises

Practice problems focused on **basic Python list operations and loops**.
Each exercise includes a **problem statement** and a **simple solution**.

Topics covered: iteration, conditions, counting, filtering, replacing values, and basic list manipulation.

## Exercise 1

**Request:**

Create a list of numbers and print each element using a for loop.

```python
numbers = [3, 7, 10, 15, 2]

for num in numbers:
    print(num)
```

## Exercise 2

**Request:**

Given a list of numbers, calculate the sum of all elements using a loop.

```python
numbers = [3, 7, 10, 15, 2]
total = 0

for num in numbers:
    total += num

print("Sum:", total)
```

## Exercise 3

**Request:**

Given a list of numbers, count how many numbers are even.

```python
numbers = [3, 7, 10, 15, 2, 8, 11]
count_even = 0

for num in numbers:
    if num % 2 == 0:
        count_even += 1

print("Even numbers:", count_even)
```

## Exercise 4

**Request:**

Given a list of numbers, find the largest number without using `max()`.

```python
numbers = [3, 7, 10, 15, 2, 8, 11]
largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print("Largest:", largest)
```

## Exercise 5

**Request:**

Given a list of numbers, replace every negative number with 0.

```python
numbers = [5, -2, 9, -7, 0, 4]
result = []

for num in numbers:
    if num < 0:
        result.append(0)
    else:
        result.append(num)

print(result)
```

## Exercise 6

**Request:**

Given a list of boolean values, count how many values are `True`.

```python
values = [True, False, True, True, False, False, True]
count_true = 0

for value in values:
    if value == True:
        count_true += 1

print("True count:", count_true)
```

## Exercise 7

**Request:**

Given a list of numbers, create a new list that contains only numbers greater than 10.

```python
numbers = [4, 12, 7, 25, 9, 18]
greater_than_10 = []

for num in numbers:
    if num > 10:
        greater_than_10.append(num)

print(greater_than_10)
```

## Exercise 8

**Request:**

Given a list of words, count how many words have more than 5 characters.

```python
words = ["apple", "banana", "kiwi", "computer", "book", "orange"]
count = 0

for word in words:
    if len(word) > 5:
        count += 1

print("Words with more than 5 characters:", count)
```

## Exercise 9

**Request:**

Given a list of numbers, check if a specific number exists in the list.

```python
numbers = [3, 7, 10, 15, 2]
target = 10
found = False

for num in numbers:
    if num == target:
        found = True
        break

print("Exists:", found)
```

## Exercise 10

**Request:**

Given a list of numbers, count how many numbers are positive, negative, and zero.

```python
numbers = [5, -2, 0, 9, -7, 0, 4, -1]
positive = 0
negative = 0
zero = 0

for num in numbers:
    if num > 0:
        positive += 1
    elif num < 0:
        negative += 1
    else:
        zero += 1

print("Positive:", positive)
print("Negative:", negative)
print("Zero:", zero)
```

## Exercise 11

**Request:**

Given a list of numbers, create a new list that removes duplicate values.

```python
numbers = [1, 2, 2, 3, 4, 4, 5, 1]
unique_numbers = []

for num in numbers:
    if num not in unique_numbers:
        unique_numbers.append(num)

print(unique_numbers)
```

## Exercise 12

**Request:**

Given a list of numbers, reverse the list using a loop (do not use `reverse()` or slicing).

```python
numbers = [1, 2, 3, 4, 5]
reversed_list = []

for num in numbers:
    reversed_list.insert(0, num)

print(reversed_list)
```

## Exercise 13

**Request:**

Given a list of numbers, check if all numbers are positive.

```python
numbers = [5, 3, 9, 1, 7]
all_positive = True

for num in numbers:
    if num <= 0:
        all_positive = False
        break

print("All positive:", all_positive)
```

## Exercise 14

**Request:**

Given a list that may contain `None` values, replace every `None` with 0.

```python
values = [5, None, 8, None, 3]
result = []

for value in values:
    if value is None:
        result.append(0)
    else:
        result.append(value)

print(result)
```

## Exercise 15

**Request:**

Given a list that contains integers and string numbers, convert all string numbers to integers.

```python
values = [1, "2", 3, "4", 5, "6"]
converted = []

for value in values:
    if type(value) == str:
        converted.append(int(value))
    else:
        converted.append(value)

print(converted)
```
