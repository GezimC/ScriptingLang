## Exercise 16

**Request:**

Create a matrix (3,3), find max value and its position.

```python
matrix = [
    [3, 7, 1],
    [9, 2, 8],
    [4, 6, 5]
]

max_value = matrix[0][0]
max_row = 0
max_col = 0

for i in range(len(matrix)):
    for j in range(len(matrix[i])):
        if matrix[i][j] > max_value:
            max_value = matrix[i][j]
            max_row = i
            max_col = j

print("Matrix:")
for row in matrix:
    print(row)

print("Max value:", max_value)
print("Position:", (max_row, max_col))
```

## Exercise 17

**Request:**

Create a list of squares from another list (long and short way).

```python
numbers = [1, 2, 3, 4, 5]

# Long way
squares_long = []
for num in numbers:
    squares_long.append(num ** 2)

# Short way
squares_short = [num ** 2 for num in numbers]

print("Long way:", squares_long)
print("Short way:", squares_short)
```

## Exercise 18

**Request:**

Reverse a list using `while`.

```python
numbers = [1, 2, 3, 4, 5]
reversed_list = []

i = len(numbers) - 1
while i >= 0:
    reversed_list.append(numbers[i])
    i -= 1

print(reversed_list)
```

## Exercise 19

**Request:**

Give access to user with password `"admin123"`, use `while` to check.

```python
correct_password = "admin123"
password = ""

while password != correct_password:
    password = input("Enter password: ")

print("Access granted")
```

## Exercise 20

**Request:**

List comprehension: find just local IPs from `ips = ["192.168.1.1", "10.0.0.1", "192.168.1.45"]`.

```python
ips = ["192.168.1.1", "10.0.0.1", "192.168.1.45"]

local_ips = [ip for ip in ips if ip.startswith("192.168.")]

print(local_ips)
```

## Exercise 21

**Request:**

Represent a Network Packet: `packet = ("192.168.1.10", "8.8.8.8", 53)`.

```python
packet = ("192.168.1.10", "8.8.8.8", 53)

source_ip = packet[0]
destination_ip = packet[1]
port = packet[2]

print("Source IP:", source_ip)
print("Destination IP:", destination_ip)
print("Port:", port)
```
